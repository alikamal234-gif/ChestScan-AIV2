```python
import numpy as np
import keras

import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

MODEL_PATH = r"c:\Users\Youcode\Desktop\ChestScan-AIV2\models\fine_tuning_model.keras"
IMAGE_PATH = r"c:\Users\Youcode\Desktop\ChestScan-AIV2\models\img.jpeg"
IMG_HEIGHT = 224
IMG_WIDTH = 224

keras.config.enable_unsafe_deserialization()
model = keras.models.load_model(
    MODEL_PATH,
    compile=False,
    safe_mode=False,
    custom_objects={"preprocess_input": preprocess_input},
)

img = keras.utils.load_img(
    IMAGE_PATH,
    target_size=(IMG_HEIGHT, IMG_WIDTH)
)

img_array = keras.utils.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)

predictions = model.predict(img_array, verbose=0)

if predictions.shape[-1] == 1:
    probability_pneumonia = float(predictions[0, 0])
    predicted_label = "PNEUMONIA" if probability_pneumonia >= 0.5 else "NORMAL"
    print(f"Raw Prediction Probabilities: {predictions}")
    print(f"Predicted Label: {predicted_label}")
    print(f"Pneumonia probability: {probability_pneumonia:.4f}")
else:
    class_names = ["NORMAL", "PNEUMONIA"]
    predicted_class_idx = int(np.argmax(predictions[0]))
    print(f"Raw Prediction Probabilities: {predictions}")
    print(f"Predicted Class Index: {predicted_class_idx}")
    print(f"Predicted Label: {class_names[predicted_class_idx]}")
```
